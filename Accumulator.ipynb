```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("AccumulatorExample").getOrCreate()
sc =spark.sparkContext
```

```python
# create accumulator
invalid_count = sc.accumulator(0)

# sample RDD data
transactions = sc.parallelize([
    {"txn_id":1, "amount":100},
    {"txn_id":2, "amount":-10},
    {"txn_id":3, "amount":200},
    {"txn_id":4, "amount":-20},
    {"txn_id":5, "amount":100},
])

# increment accumulator
def validate(txn):
    global invalid_count
    if txn["amount"] < 0:
        invalid_count += 1
    return txn

transactions.map(validate).collect()

print(f"Number of invalid transactions: {invalid_count.value}")
```

```python
# increment accumulator, global value so count increases
def validate_id(txn):
    global invalid_count
    if txn["txn_id"]%2 == 1:
        invalid_count += 1
    return txn
transactions.map(validate_id).collect()

print(f"Number of invalid transactions: {invalid_count.value}")
```
