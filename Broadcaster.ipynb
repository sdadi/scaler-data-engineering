```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("BroadCasterExample").getOrCreate()
sc =spark.sparkContext
```

```python
product_catalog = {
    1:"Laptop",
    2: "Phone",
    3: "Tablet"
}

broadcast_catalog =sc.broadcast(product_catalog)

transactions = sc.parallelize([
    {"txn_id":101, "product_id":1},
    {"txn_id":102, "product_id":2},
    {"txn_id":103, "product_id":3},
])

def enrich(txn):
    txn['product_name'] = broadcast_catalog.value.get(txn['product_id'],'Unknown')
    return txn

enriched_transactions = transactions.map(enrich).collect()
print(enriched_transactions)
```
